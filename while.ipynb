## 1. Что такое цикл `while`?

**Цикл `while`** (цикл с предусловием) — это конструкция, которая **повторяет** блок команд до тех пор, пока условие остаётся **истинным** (`True`).

Особенность: условие проверяется **до** выполнения тела цикла (поэтому «с предусловием»). Если условие сразу ложно — тело не выполнится ни разу.

**Когда использовать:**
- Когда заранее неизвестно, сколько раз нужно повторить действие.
- Когда повторение зависит от некоторого условия (например, пока пользователь не введёт правильный пароль).

> ⚠️ Важно: внутри тела цикла должно быть действие, которое **приближает** условие к `False`, иначе цикл станет **бесконечным**.

## 2. Синтаксис и схема работы

```python
while условие:
    # тело цикла (команды с отступом)
    ...
```

**Схема работы:**
1. Проверяется `условие`.
2. Если `True` → выполняется тело цикла → снова к шагу 1.
3. Если `False` → цикл завершается, управление переходит к командам после цикла.

**Схема (текстовая):**
```
      ┌──────────────┐
      │   условие?   │
      └──────┬───────┘
       True  │  False
      ┌──────▼───────┐
      │  тело цикла  │──┐
      └──────────────┘  │
             ▲──────────┘
             │
      ┌──────▼───────┐
      │ после цикла  │
      └──────────────┘
```

## 3. Пример 1: Счётчик от 1 до 5

Выведем числа 1, 2, 3, 4, 5. Здесь условие — `i <= 5`.

````python
i = 1                 # начальное значение счётчика
while i <= 5:         # условие продолжения
    print(i)
    i = i + 1         # увеличиваем счётчик (приближаем условие к False)

print("Цикл завершён")
````

**Разбор:**
- `i = 1` — инициализация.
- Проверка `1 <= 5` → True → печать `1` → `i = 2`.
- ... и так до `i = 5`. После печати `5` получаем `i = 6`.
- Проверка `6 <= 5` → False → выход из цикла.

💡 Если забыть строку `i = i + 1`, получится **бесконечный цикл** (печать `1` без остановки).

## 4. Пример 2: Сумма чисел от 1 до N

Пользователь вводит `N`, программа считает `1 + 2 + ... + N`.

````python
N = int(input("Введите N: "))

i = 1
total = 0
while i <= N:
    total = total + i
    i = i + 1

print("Сумма от 1 до", N, "равна", total)
````

**Разбор:**
- `total` — накопитель суммы, начинаем с 0.
- На каждом шаге добавляем текущее `i` и увеличиваем `i`.
- Когда `i` станет `N + 1`, условие станет ложным — цикл остановится.

**Проверка:** для `N = 5` должно получиться `15` (1+2+3+4+5).

## 5. Пример 3: Ввод пароля (интерактивный)

Цикл `while` идеален, когда число попыток **заранее неизвестно**.

````python
password = ""
while password != "python123":
    password = input("Введите пароль: ")
    if password != "python123":
        print("Неверно, попробуйте ещё раз")

print("Доступ разрешён!")
````

**Разбор:**
- До цикла `password` пустая — условие `"" != "python123"` истинно, входим в цикл.
- Внутри запрашиваем ввод. Если неверно — сообщение и снова цикл.
- Как только введён правильный пароль — условие ложно, выход.

> 💡 В Google Colab `input()` работает через специальное поле. В обычном Jupyter — через диалог.

## 6. Пример 4: Обработка цифр числа

Найдём сумму цифр числа `n` (например, 1234 → 1+2+3+4 = 10). Используем `n % 10` (последняя цифра) и `n // 10` (отбрасываем её).

````python
n = 1234
s = 0
while n > 0:
    digit = n % 10
    s = s + digit
    n = n // 10

print("Сумма цифр:", s)
````

**Разбор по шагам (n = 1234):**

| Шаг | n | digit = n % 10 | s | n // 10 |
|-----|---|----------------|---|---------|
| 1 | 1234 | 4 | 4 | 123 |
| 2 | 123 | 3 | 7 | 12 |
| 3 | 12 | 2 | 9 | 1 |
| 4 | 1 | 1 | 10 | 0 |

Когда `n = 0`, условие `n > 0` ложно — выход. Результат: **10**.

## 7. Пример 5: Игра «Угадай число»

Программа загадывает число, пользователь угадывает. Цикл продолжается, пока не угадает.

````python
import random

secret = random.randint(1, 10)
guess = 0
attempts = 0

while guess != secret:
    guess = int(input("Угадайте число от 1 до 10: "))
    attempts = attempts + 1
    if guess < secret:
        print("Больше")
    elif guess > secret:
        print("Меньше")

print("Угадали! Попыток:", attempts)
````

**Разбор:**
- `guess = 0` — начальное значение, заведомо не равное загаданному (1..10).
- Внутри цикла запрашиваем ввод и сравниваем с `secret`.
- Цикл завершается, как только `guess == secret`.

**Альтернатива:** можно использовать `while True` + `break`:

````python
# Вариант с while True и break
secret = random.randint(1, 10)
attempts = 0
while True:
    guess = int(input("Угадайте число от 1 до 10: "))
    attempts += 1
    if guess == secret:
        print("Угадали! Попыток:", attempts)
        break
    elif guess < secret:
        print("Больше")
    else:
        print("Меньше")
````

## 8. Типичные ошибки

### ❌ Ошибка 1: Забыли изменить условие (бесконечный цикл)
```python
i = 1
while i <= 5:
    print(i)      # i не меняется — печать 1 вечно
```
**Как остановить:** в Jupyter — кнопка ⏹ (Stop/Interrupt Kernel).

### ❌ Ошибка 2: Неверное начальное значение
```python
i = 10
while i <= 5:
    print(i)      # условие сразу False, тело не выполнится ни разу
    i += 1
```

### ❌ Ошибка 3: Забыли отступ
```python
while i <= 5:
print(i)          # IndentationError
```

### ❌ Ошибка 4: Сравнение вместо присваивания
```python
while i <= 5:
    i == i + 1    # это сравнение, а не изменение! Бесконечный цикл
```
Правильно: `i = i + 1` или `i += 1`.

### ✅ Полезные операторы
- `break` — досрочно выйти из цикла.
- `continue` — перейти к следующей итерации, пропустив остаток тела.
- `else` у цикла — выполняется, если цикл завершился **без** `break`.

````python
# Демонстрация break и continue
i = 0
while i < 10:
    i += 1
    if i == 3:
        continue      # пропустить печать 3
    if i == 7:
        break         # остановиться на 7
    print(i)
else:
    print("Цикл завершился без break")

print("Конец")
````

**Вывод:** 1, 2, 4, 5, 6, затем «Конец». Строка `else` **не** напечатается, т.к. сработал `break`.
